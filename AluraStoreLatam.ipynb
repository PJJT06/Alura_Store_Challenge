### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



###**En este primer análisis, debes calcular el ingreso total de cada tienda. Sumando los valores de la columna Precio de cada conjunto de datos de la tienda para estimar los ingresos.**

```python
Ingreso_tienda = tienda["Precio"].sum()
Ingreso_tienda
```

```python
Ingreso_Tienda2 = tienda2["Precio"].sum()
Ingreso_Tienda2
```

```python
Ingreso_tienda3 = tienda3["Precio"].sum()
Ingreso_tienda3
```

```python
Ingreso_tiend4 = tienda4["Precio"].sum()
Ingreso_tiend4
```

## **Generando Graficos**

### Preparar datos para gráficos



```python
data_ingresos = {
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Ingresos Totales': [Ingreso_tienda, Ingreso_Tienda2, Ingreso_tienda3, Ingreso_tiend4]
}
df_ingresos = pd.DataFrame(data_ingresos)
print("Ingresos Totales por Tienda:")
print(df_ingresos)
```

**Reasoning**:
Now I will extract the top 3 most sold product categories for each store and consolidate them into a single DataFrame called `df_ventas_categoria`. This DataFrame will include the store name, product category, and the count of sales for each category.



```python
data_ventas_categoria = []

#Tienda 1
for category, count in vendidos_tienda.head(3).items():
    data_ventas_categoria.append({'Tienda': 'Tienda 1', 'Categoría del Producto': category, 'Cantidad de Ventas': count})

#Tienda 2
for category, count in vendidos_tienda2.head(3).items():
    data_ventas_categoria.append({'Tienda': 'Tienda 2', 'Categoría del Producto': category, 'Cantidad de Ventas': count})

#Tienda 3
for category, count in vendidos_tienda3.head(3).items():
    data_ventas_categoria.append({'Tienda': 'Tienda 3', 'Categoría del Producto': category, 'Cantidad de Ventas': count})

#Tienda 4
for category, count in vendidos_tienda4.head(3).items():
    data_ventas_categoria.append({'Tienda': 'Tienda 4', 'Categoría del Producto': category, 'Cantidad de Ventas': count})

df_ventas_categoria = pd.DataFrame(data_ventas_categoria)
print("Top 3 Ventas por Categoría por Tienda:")
print(df_ventas_categoria)
```

```python
data_calificaciones = {
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Calificación Promedio': [calificacion_tienda, calificacion_tienda2, calificacion_tienda3, calificacion_tienda4]
}
df_calificaciones = pd.DataFrame(data_calificaciones)
print("Calificaciones Promedio por Tienda:")
print(df_calificaciones)
```

## Graficar Ingresos Totales por Tienda (Gráfico de Barras)




```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(df_ingresos['Tienda'], df_ingresos['Ingresos Totales'], color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])
plt.title('Ingresos Totales por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingresos Totales')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## Graficar Ventas por Categoría por Tienda


```python
import matplotlib.pyplot as plt

stores = df_ventas_categoria['Tienda'].unique()

for store in stores:
    df_store = df_ventas_categoria[df_ventas_categoria['Tienda'] == store]

    plt.figure(figsize=(10, 6))
    plt.barh(df_store['Categoría del Producto'], df_store['Cantidad de Ventas'], color='lightcoral')
    plt.xlabel('Cantidad de Ventas')
    plt.ylabel('Categoría del Producto')
    plt.title(f'Top 3 Categorías de Ventas en {store}')
    plt.gca().invert_yaxis()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()
```

## Graficar Calificación Promedio por Tienda


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(df_calificaciones['Tienda'], df_calificaciones['Calificación Promedio'], color='purple', s=100)
plt.title('Calificación Promedio por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Calificación Promedio')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(3.5, 4.5) # Adjust y-axis limits for better visualization
plt.show()
```

En conclusión, la Tienda 4 es la opción más adecuada para ser vendida, ya que es la que menor impacto negativo tendría sobre el desempeño financiero global del conjunto de empresas, permitiendo fortalecer y potenciar aquellas que presentan mejores resultados.

### Mapa de Calor de Ubicaciones de Compra

```python
import folium
from folium.plugins import HeatMap


all_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4])

# Obtener el centro del mapa
center_lat = all_tiendas['lat'].mean()
center_lon = all_tiendas['lon'].mean()

# Crear un mapa base de Folium
m = folium.Map(location=[center_lat, center_lon], zoom_start=6)

# Añadir una capa de calor
HeatMap(data=all_tiendas[['lat', 'lon']].dropna().values.tolist()).add_to(m)

# Mostrar el mapa
m
```
